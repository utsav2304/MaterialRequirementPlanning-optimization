```python
merged=bom.merge(production_plan[["product_id","planned_qty","due_date"]],on="product_id")
```

```python
merged["gross_quan"]=merged["qty_per_unit"] * merged["planned_qty"]
merged.head(3)
```

```python
gross_by_material=merged.groupby("material_id")["gross_quan"].sum().reset_index()
```

```python
earliest_due = merged.groupby("material_id")["due_date"].min().reset_index()
earliest_due.rename(columns={"due_date":"earliest_due_date"}, inplace=True)
```

```python
mrp=inventory.merge(gross_by_material,on="material_id", how="left")
```

```python
mrp=mrp.merge(earliest_due,on="material_id",how="left")
```

```python
mrp["gross_quan"] = mrp["gross_quan"].fillna(0)
mrp["earliest_due_date"]=mrp["earliest_due_date"].fillna(0)
```

```python
mrp["safety_stock_qty"]=mrp["safety_stock_pct"]*mrp["current_stock"]
```

```python
mrp["net_requirement"]=(mrp["safety_stock_qty"]+mrp["gross_quan"]-mrp["current_stock"]).clip(lower=0)
```

```python
print(mrp["earliest_due_date"].dtype)
```

```python
mrp.info()
```

```python
mrp["earliest_due_date"]=pd.to_datetime(mrp["earliest_due_date"])
```

```python
mrp["days_available"]=(mrp["earliest_due_date"]-Today).dt.days
```

```python
import math

eoq_list = []
for index, row in mrp.iterrows():
    demand = row["gross_quan"]
    ordering_cost = row["ordering_cost_per_order"]
    holding_cost = row["holding_cost_per_unit"]

    if demand > 0 and holding_cost > 0:
        eoq = math.sqrt((2 * demand * ordering_cost) / holding_cost)
    else:
        eoq = 0
    eoq_list.append(round(eoq, 0))

mrp["eoq"] = eoq_list
```

```python
mrp["orders_per_period_eoq"] = mrp.apply(
    lambda r: round(r["gross_quan"] / r["eoq"], 1) if r["eoq"] > 0 else 0, axis=1
)
mrp.head(5)
```

```python
mrp["eoq_total_cost"] = (
    mrp["orders_per_period_eoq"] * mrp["ordering_cost_per_order"]
    + (mrp["eoq"] / 2) * mrp["holding_cost_per_unit"]
).round(2)

mrp.head(5)
```

```python
mrp["single_order_total_cost"] = mrp.apply(
    lambda r: r["ordering_cost_per_order"] + (r["gross_quan"] / 2) * r["holding_cost_per_unit"]
    if r["gross_quan"] > 0 else 0, axis=1
).round(2)
```

```python
mrp["eoq_savings_vs_single_order"] = (mrp["single_order_total_cost"] - mrp["eoq_total_cost"]).round(2)
```

```python
# Pulp
```

```python
shortage_materials = mrp[mrp["net_requirement"] > 0].copy()
```

```python
allocation_rows = []
unmet_rows = []
```

```python
for _, mat_row in shortage_materials.iterrows():
    mat_id = mat_row["material_id"]
    need_qty = mat_row["net_requirement"]
    days_avail = mat_row["days_available"]
    mat_suppliers = suppliers[suppliers["material_id"] == mat_id].copy()
    mat_suppliers = mat_suppliers[mat_suppliers["lead_time_days"] <= days_avail]

    if mat_suppliers.empty:
        unmet_rows.append({
            "material_id": mat_id,
            "material_name": mat_row["material_name"],
            "net_requirement": need_qty,
            "reason": "No supplier can meet lead time before due date"
        })
        continue

    mat_suppliers = mat_suppliers.reset_index(drop=True)
    n = len(mat_suppliers)

    order_qty = [pulp.LpVariable(f"qty_{mat_id}_{i}", lowBound=0) for i in range(n)]
    use_supplier = [pulp.LpVariable(f"use_{mat_id}_{i}", cat="Binary") for i in range(n)]

    prob = pulp.LpProblem(f"procure_{mat_id}", pulp.LpMinimize)   # <-- you're missing this line entirely!

    prob += pulp.lpSum([order_qty[i] * mat_suppliers.loc[i, "price_per_unit"] for i in range(n)])
    prob += pulp.lpSum([order_qty[i] for i in range(n)]) >= need_qty

    for i in range(n):
        moq = mat_suppliers.loc[i, "moq"]
        cap = mat_suppliers.loc[i, "max_capacity"]
        prob += order_qty[i] >= moq * use_supplier[i]
        prob += order_qty[i] <= cap * use_supplier[i]

    prob.solve(pulp.PULP_CBC_CMD(msg=0))

    if pulp.LpStatus[prob.status] != "Optimal":
        unmet_rows.append({
            "material_id": mat_id,
            "material_name": mat_row["material_name"],
            "net_requirement": need_qty,
            "reason": "No feasible allocation within MOQ/capacity constraints"
        })
    else:
        for i in range(n):
            qty = order_qty[i].varValue
            if qty and qty > 0.01:
                allocation_rows.append({
                    "material_id": mat_id,
                    "material_name": mat_row["material_name"],
                    "supplier_name": mat_suppliers.loc[i, "supplier_name"],
                    "order_qty": round(qty, 2),
                    "price_per_unit": mat_suppliers.loc[i, "price_per_unit"],
                    "total_cost": round(qty * mat_suppliers.loc[i, "price_per_unit"], 2),
                    "lead_time_days": mat_suppliers.loc[i, "lead_time_days"]
                })
```

```python
allocation_df = pd.DataFrame(allocation_rows)
unmet_df = pd.DataFrame(unmet_rows)

total_procurement_cost = allocation_df["total_cost"].sum()
materials_fully_sourced = allocation_df["material_id"].nunique()
materials_unmet = unmet_df["material_id"].nunique() if not unmet_df.empty else 0
total_materials_short = len(shortage_materials)

print(f"Total procurement spend: ₹{total_procurement_cost:,.2f}")
print(f"Materials sourced: {materials_fully_sourced} / {total_materials_short}")
print(f"Materials unmet: {materials_unmet}")
```
